# example package Tutorial

This package only has two modules: a module to manually do a discrete convolution of a point spread function (psf) matrix K and a real image f and a module to load example data. The convolution module includes two functions, one for scalar images f and one for vector images f which has not been implemented yet.



```python
import numpy as np
from example import convolution
```

Run scalar convolution with some inputs:

```python
psf_K = np.array([0, 1, 23, 8, 17, -8, -40])
image_f = np.pi
print(convolution.scalar(psf_K, image_f))
```

Running vector convolution will return an error:

```python
psf_K = np.array([0, 1, 23, 8, 17, -8, -40])
image_f = np.pi
print(convolution.vector(psf_K, image_f))
```
